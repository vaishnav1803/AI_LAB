```python
import heapq

def calculate_manhattan(state, goal):

    total_distance = 0
    for tile in range(1, 9):
        curr_idx = state.index(tile)
        goal_idx = goal.index(tile)

        curr_row, curr_col = curr_idx // 3, curr_idx % 3
        goal_row, goal_col = goal_idx // 3, goal_idx % 3

        total_distance += abs(curr_row - goal_row) + abs(curr_col - goal_col)

    return total_distance


def solve_8_puzzle_manhattan(start, goal):

    def get_moves(state):
        moves = []
        blank_idx = state.index(0)
        row, col = blank_idx // 3, blank_idx % 3
        directions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for dr, dc in directions:
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_idx = new_row * 3 + new_col
                new_state = list(state)
                new_state[blank_idx], new_state[new_idx] = new_state[new_idx], new_state[blank_idx]
                moves.append(tuple(new_state))
        return moves

    h_start = calculate_manhattan(start, goal)
    queue = [(h_start, 0, start, [start])]
    visited = set()

    while queue:
        f, g, current, path = heapq.heappop(queue)

        if current == goal:
            return path

        if current in visited:
            continue
        visited.add(current)

        for next_state in get_moves(current):
            if next_state not in visited:
                g_new = g + 1
                h_new = calculate_manhattan(next_state, goal)
                heapq.heappush(queue, (g_new + h_new, g_new, next_state, path + [next_state]))

    return None


def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()

start_board = (2, 8, 3,
               1, 6, 4,
               7, 0, 5)

goal_board  = (1, 2, 3,
               8, 0, 4,
               7, 6, 5)

solution_path = solve_8_puzzle_manhattan(start_board, goal_board)

if solution_path:
    print(f"Solved using Manhattan Distance in {len(solution_path) - 1} moves!\n")
    for step, state in enumerate(solution_path):
        print(f"Step {step}:")
        print_board(state)
else:
    print("This puzzle setup is unsolvable.")
```
