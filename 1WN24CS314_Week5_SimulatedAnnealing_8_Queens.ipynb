```python
import math
import random


def generate_random_board(n):

    return [random.randint(0, n - 1) for _ in range(n)]


def calculate_cost(board):
    attacks = 0
    n = len(board)
    for c1 in range(n):
        for c2 in range(c1 + 1, n):
            if board[c1] == board[c2]:
                attacks += 1
            elif abs(board[c1] - board[c2]) == abs(c1 - c2):
                attacks += 1
    return attacks


def get_neighbor(current_board, n):
    neighbor = list(current_board)
    random_col = random.randint(0, n - 1)

    new_row = random.randint(0, n - 1)
    while new_row == neighbor[random_col]:
        new_row = random.randint(0, n - 1)

    neighbor[random_col] = new_row
    return neighbor


def solve_n_queens_sa(
    n, initial_temp=100.0, cooling_rate=0.995, max_steps=10000
):
    """Solves N-Queens using Simulated Annealing logic from the image."""
    current_state = generate_random_board(n)
    current_cost = calculate_cost(current_state)

    t = initial_temp
    steps = 0

    while t > 0.001 and steps < max_steps:
        if current_cost == 0:
            break

        neighbor_state = get_neighbor(current_state, n)
        neighbor_cost = calculate_cost(neighbor_state)

        delta_e = current_cost - neighbor_cost

        if delta_e > 0:
            current_state = neighbor_state
            current_cost = neighbor_cost
        else:
            probability = math.exp(delta_e / t)
            if random.random() < probability:
                current_state = neighbor_state
                current_cost = neighbor_cost

        t *= cooling_rate
        steps += 1

    return current_state, current_cost, steps


if __name__ == "__main__":
    N = 8
    print(f"Solving {N}-Queens using Simulated Annealing...")

    solution, final_cost, completed_steps = solve_n_queens_sa(N)

    print("\n--- Execution Details ---")
    print(f"Steps Taken: {completed_steps}")
    print(f"Final Conflicts (Cost): {final_cost}")
    print(f"Solution Array: {solution}")

    if final_cost == 0:
        print("\nSuccess! Found a valid board configuration:")
        for row in range(N):
            row_str = ""
            for col in range(N):
                row_str += " Q " if solution[col] == row else " . "
            print(row_str)
    else:
        print("\nCould not find a perfect solution within parameters.")
```
