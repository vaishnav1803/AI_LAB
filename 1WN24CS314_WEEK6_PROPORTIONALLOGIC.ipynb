```python
import itertools


def implies(p, q):
    return not p or q


def check_kb(p, q, r):
    s1 = implies(q, p)
    s2 = implies(p, not q)
    s3 = q or r
    return s1, s2, s3, (s1 and s2 and s3)


print(f"{'P':<6}{'Q':<6}{'R':<6}{'Q->P':<8}{'P->~Q':<8}{'QvR':<8}{'KB':<6}")
print("-" * 50)

all_combinations = list(itertools.product([True, False], repeat=3))
models = []

for p, q, r in all_combinations:
    s1, s2, s3, kb = check_kb(p, q, r)


    p_str, q_str, r_str = str(p)[0], str(q)[0], str(r)[0]
    s1_str, s2_str, s3_str, kb_str = str(s1)[0], str(s2)[0], str(s3)[0], str(kb)[0]


    model_marker = f" <-- Model found!" if kb else ""
    print(f"{p_str:<6}{q_str:<6}{r_str:<6}{s1_str:<8}{s2_str:<8}{s3_str:<8}{kb_str:<6}{model_marker}")

    if kb:
        models.append({'P': p, 'Q': q, 'R': r})

print("\n" + "="*50)
print(f"Total satisfying models found: {len(models)}")
for idx, model in enumerate(models, 1):
    print(f"  Model {idx}: P={model['P']}, Q={model['Q']}, R={model['R']}")
print("="*50 + "\n")




entails_r = all(model['R'] == True for model in models)
print(f"ii) Does KB entail R? {entails_r}")

entails_r_implies_p = all(implies(model['R'], model['P']) == True for model in models)
print(f"iii) Does KB entail R -> P? {entails_r_implies_p}")


entails_q_implies_r = all(implies(model['Q'], model['R']) == True for model in models)
print(f"iv) Does KB entail Q -> R? {entails_q_implies_r}")
```
