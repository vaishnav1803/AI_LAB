```python
def is_variable(expr):
    return isinstance(expr, str) and expr.islower()

def is_constant(expr):
    return isinstance(expr, str) and expr.isupper()

def is_compound(expr):
    return isinstance(expr, tuple) or isinstance(expr, list)

def occur_check(var, expr, theta):
    if var == expr:
        return True
    if is_variable(expr) and expr in theta:
        return occur_check(var, theta[expr], theta)
    if is_compound(expr):
        return any(occur_check(var, arg, theta) for arg in expr[1:])
    return False

def unify_variable(var, expr, theta):
    if var in theta:
        return unify(theta[var], expr, theta)
    elif is_variable(expr) and expr in theta:
        return unify(var, theta[expr], theta)
    elif occur_check(var, expr, theta):
        return None
    else:
        new_theta = theta.copy()
        new_theta[var] = expr
        return new_theta

def unify(x, y, theta=None):
    if theta is None:
        theta = {}

    if theta is None:
        return None

    if x == y:
        return theta

    if is_variable(x):
        return unify_variable(x, y, theta)

    if is_variable(y):
        return unify_variable(y, x, theta)

    if is_compound(x) and is_compound(y):
        if len(x) != len(y) or x[0] != y[0]:
            return None

        for arg_x, arg_y in zip(x[1:], y[1:]):
            theta = unify(arg_x, arg_y, theta)
            if theta is None:
                return None
        return theta

    return None

if __name__ == "__main__":
    test_cases = [
        (('King', 'x'), ('King', 'John')),
        (('P', 'x', 'y'), ('P', 'a', 'z')),

        (('King', 'x'), ('Queen', 'y'))
    ]

    for expr1, expr2 in test_cases:
        result = unify(expr1, expr2)
        print(f"Unify({expr1}, {expr2}) -> {result}")
```
