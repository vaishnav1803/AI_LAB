```python
import random
import copy

def calculate_conflicts(board):
    n = len(board)
    conflicts = 0
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j]:
                conflicts += 1
            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def generate_random_board(n):
    return [random.randint(0, n - 1) for _ in range(n)]

def pure_hill_climbing(n):
    current_board = generate_random_board(n)
    current_conflicts = calculate_conflicts(current_board)
    steps = 0

    while True:
        if current_conflicts == 0:
            return current_board, current_conflicts, steps, "SUCCESS (Global Optimum Found)"

        next_board = None
        min_neighbor_conflicts = current_conflicts

        for col in range(n):
            original_row = current_board[col]
            for row in range(n):
                if row != original_row:
                    temp_board = copy.deepcopy(current_board)
                    temp_board[col] = row
                    temp_conflicts = calculate_conflicts(temp_board)

                    if temp_conflicts < min_neighbor_conflicts:
                        min_neighbor_conflicts = temp_conflicts
                        next_board = temp_board

        if next_board is not None:
            current_board = next_board
            current_conflicts = min_neighbor_conflicts
            steps += 1
        else:
            return current_board, current_conflicts, steps, "FAILURE (Stuck in a Local Minimum)"

if __name__ == "__main__":
    print("--- Pure Hill Climbing N-Queens Solver (No Restarts) ---")

    try:
        n_input = int(input("Enter the number of queens (N): "))
    except ValueError:
        n_input = 8
        print("Invalid input. Defaulting to N = 8.")

    seed_input = input("Enter the start state ").strip()

    if seed_input:
        try:
            random.seed(int(seed_input))
            print(f"Random state initialized with seed: {seed_input}")
        except ValueError:
            random.seed(seed_input)
            print(f"Random state initialized with string seed: '{seed_input}'")
    else:
        print("Running with dynamic random state.")

    print("\nExecuting pure hill climbing...")
    final_board, remaining_conflicts, total_steps, status = pure_hill_climbing(n_input)

    print(f"\n--- Results Summary ---")
    print(f"Status: {status}")
    print(f"Total Steps Taken: {total_steps}")
    print(f"Final Conflicts Remaining: {remaining_conflicts}")
    print(f"Board State (Index=Col, Value=Row): {final_board}")

    print("\nFinal Board Layout:")
    for row in range(n_input):
        line = ""
        for col in range(n_input):
            if final_board[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
```
